### Model Implementation

```python
### Seperate Features  and Target
target=df['Y']
features=df.drop('Y',axis=1)
```

```python
### Encoding
lab_en=LabelEncoder()
target=lab_en.fit_transform(target)
```

```python
### Train_Test_Split
x_train, x_test, y_train, y_test = train_test_split(features,target,test_size=0.2,random_state=100,stratify=target)

print("Training Features:", x_train.shape)
print("Testing Features :", x_test.shape)
print("Training Target  :", y_train.shape)
print("Testing Target   :", y_test.shape)
```

```python
### Scaling
std_sca = StandardScaler()

x_train = std_sca.fit_transform(x_train)
x_test = std_sca.transform(x_test)
```

* The input features were normalized using StandardScaler.
* The scaler was fitted on the training data and then applied to the test data.
* This helps the ANN converge faster and improves model performance.

```python
### Basic ANN Model Building

model = Sequential()

# Input layer + Hidden layer
model.add(Dense(32, input_dim=60, activation='relu'))

# Second hidden layer
model.add(Dense(16, activation='relu'))

# Output layer
model.add(Dense(1, activation='sigmoid'))
```

```python
### Model Compilation
model.compile(optimizer='adam',loss='binary_crossentropy',metrics=['accuracy'])
```

```python
model.summary()
```

* The ANN model was successfully constructed using TensorFlow/Keras.
* The model contains two hidden layers with 32 and 16 neurons using the ReLU activation function.
* The output layer contains 1 neuron with the Sigmoid activation function for binary classification.
* The model has a total of 2,497 trainable parameters.

```python
### Model Training and validation
history = model.fit(x_train, y_train, epochs=100, batch_size=1000)
```

```python
### Prediction
y_pred_prob = model.predict(x_test)

y_pred = (y_pred_prob > 0.5).astype(int)
```

```python
### Evaluation
print("Accuracy :", accuracy_score(y_test, y_pred))
print()
print("classification Report")
print(classification_report(y_test, y_pred))
```

* The model achieved an accuracy of 76.19% on the test data.
* The precision is 75%, indicating that 75% of the samples predicted as mines were correctly classified.
* The recall is 75%, meaning the model correctly identified 75% of the actual positive samples.
* The F1-score is 75%, showing a balanced performance between precision and recall.
* Overall, the basic ANN model provides a reasonable classification performance for distinguishing between mines and rocks.

### Hyperparameter Tuning

```python
!pip install keras-tuner
```

```python
def build_model(hp):

    model = Sequential()

    # First Hidden Layer
    model.add(Dense(
        units=hp.Int('units1', min_value=16, max_value=64, step=16),
        activation=hp.Choice('activation1', ['relu', 'tanh']),
        input_shape=(60,)
    ))

    # Second Hidden Layer
    model.add(Dense(
        units=hp.Int('units2', min_value=8, max_value=32, step=8),
        activation=hp.Choice('activation2', ['relu', 'tanh'])
    ))

    # Output Layer
    model.add(Dense(1, activation='sigmoid'))

    model.compile(
        optimizer=Adam(
            learning_rate=hp.Choice(
                'learning_rate',
                values=[0.01, 0.001, 0.0001]
            )
        ),
        loss='binary_crossentropy',
        metrics=['accuracy']
    )

    return model
```

```python
import keras_tuner as kt
tuner = kt.RandomSearch(
    hypermodel=build_model,
    objective='val_accuracy',
    max_trials=10,
    overwrite=True,
    directory='random_search',
    project_name='sonar_ann'
)
```

```python
tuner.search(
    x_train,
    y_train,
    epochs=50,
    validation_split=0.2,
    verbose=1
)
```

```python
best_model = tuner.get_best_models(num_models=1)[0]
```

```python
best_hp = tuner.get_best_hyperparameters(num_trials=1)[0]

print("Units in Hidden Layer 1:", best_hp.get('units1'))
print("Activation 1:", best_hp.get('activation1'))
print("Units in Hidden Layer 2:", best_hp.get('units2'))
print("Activation 2:", best_hp.get('activation2'))
print("Learning Rate:", best_hp.get('learning_rate'))
```

```python
y_pred_prob = best_model.predict(x_test)
y_pred = (y_pred_prob > 0.5).astype(int)
```

### Evaluation

```python
print("Accuracy :", accuracy_score(y_test, y_pred))
print()
print("classification Report")
print(classification_report(y_test, y_pred))
```

* Random Search was used to optimize the ANN hyperparameters.
* The best model used:
* Hidden Layer 1: 64 neurons
* Activation Function 1: tanh
* Hidden Layer 2: 32 neurons
* Activation Function 2: tanh
* Learning Rate: 0.01
* The highest validation accuracy obtained during tuning was 88.24%.
* When evaluated on the test dataset, the tuned model achieved an accuracy of 76.19%.
